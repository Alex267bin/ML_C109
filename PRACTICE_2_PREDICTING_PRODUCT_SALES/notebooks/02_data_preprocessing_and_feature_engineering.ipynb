# 02 – Data Preprocessing and Feature Engineering
The project uses only fields available in the selected dataset. Missing numeric values are imputed inside the model pipeline to avoid fitting preprocessing on validation data.


```python
from pathlib import Path
import pandas as pd
DATA = Path('../data/raw') if Path('../data/raw/train.csv').exists() else Path('data/raw')
train = pd.read_csv(DATA / 'train.csv', parse_dates=['date'])
stores = pd.read_csv(DATA / 'stores.csv')
holidays = pd.read_csv(DATA / 'holidays_events.csv', parse_dates=['date'])
oil = pd.read_csv(DATA / 'oil.csv', parse_dates=['date'])
```

```python
data = train.merge(stores, on='store_nbr', how='left').merge(oil, on='date', how='left')
holidays['is_holiday'] = ~holidays['type'].isin(['Work Day'])
holiday_daily = holidays.groupby('date', as_index=False)['is_holiday'].max()
data = data.merge(holiday_daily, on='date', how='left')
data['is_holiday'] = data['is_holiday'].fillna(False).astype(int)
data['year'] = data.date.dt.year
data['month'] = data.date.dt.month
data['day'] = data.date.dt.day
data['day_of_week'] = data.date.dt.dayofweek
data.head()
```

# 01 – Exploratory Data Analysis
This notebook profiles the Favorita training data. Run it from the project directory after placing the Kaggle CSV files in `data/raw/`.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

DATA = Path('../data/raw') if Path('../data/raw/train.csv').exists() else Path('data/raw')
train = pd.read_csv(DATA / 'train.csv', parse_dates=['date'])
train.head()
```

```python
train.info()
train.describe(include='all').T
```

```python
train.isna().sum().sort_values(ascending=False)
```

```python
monthly_sales = train.groupby(train['date'].dt.to_period('M'))['sales'].sum()
monthly_sales.plot(figsize=(12, 5), title='Monthly total sales');
plt.xlabel('Month'); plt.ylabel('Sales'); plt.show()
```

```python
train.groupby('family')['sales'].sum().sort_values(ascending=False).head(15).sort_values().plot(kind='barh', figsize=(10, 7), title='Top product families');
plt.xlabel('Total sales'); plt.show()
```
